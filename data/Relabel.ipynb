```python
import pandas as pd
```

```python
path = '/cleaned_articles_dataset.csv'
```

```python
df = pd.read_csv(path)
```

```python
domain_counts = df.groupby(['source_domain', 'label_3']).size().reset_index(name='count')

domain_counts = domain_counts.sort_values(by='count', ascending=False)

print(domain_counts)
```

```python
label_counts = df['label_3'].value_counts().reset_index()
print(label_counts)
```

```python
df.loc[df['source_domain'] == 'www.mmnews.de', 'label_3'] = 2
```

```python
df.loc[df['source_domain'] == 'www.focus.de', 'label_3'] = 2
```

```python
df.loc[df['source_domain'] == 'www.tagesspiegel.de', 'label_3'] = 1
```

```python
label_counts = df['label_3'].value_counts().reset_index()
print(label_counts)
```
